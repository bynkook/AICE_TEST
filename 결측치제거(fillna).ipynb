### 시험문제에서 종종 결측치를 제외하고 평균값을 구해서 그 값으로 결측치를 대치하라는 문제가 나온다
- 그런데, 평균값 계산할때 pandas 는 결측치를 제외하고 평균값을 계산한다.
- 따라서 아래 첫번째와 두번째 계산은 같은 값을 출력한다.
- (주의) 다른 컬럼에도 결측치가 있다면 세번째 계산과 같이 평균값이 달라지므로 주의한다.

```python
print(df)
B_mean_1 = df.B.dropna().mean()
B_mean_2 = df.B.mean()
B_mean_3 = df.dropna()['B'].mean()
print(B_mean_1, B_mean_2)
print(B_mean_3)
```
